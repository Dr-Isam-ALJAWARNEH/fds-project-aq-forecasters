```python
# forecast_analysis.ipynb (Python cell code snippets)

import pandas as pd
from forecasting import ARIMAForecastingPipeline
from eda import GeohashForecastVisualizer

# Load data
newyork_df = pd.read_csv("../data/cleaned/newyork_cleaned.csv", parse_dates=['time'])
boston_df = pd.read_csv("../data/cleaned/boston_cleaned.csv", parse_dates=['time'])

# Ensure time column is index if needed
newyork_df.set_index("time", inplace=True)
boston_df.set_index("time", inplace=True)

# --- Run Forecasting Pipelines ---
pipeline_ny = ARIMAForecastingPipeline(newyork_df, target_col='pm25', city_name='NewYork')
pipeline_ny.run()

pipeline_boston = ARIMAForecastingPipeline(boston_df, target_col='pm25', city_name='Boston')
pipeline_boston.run()

# --- Visualize Geohash Forecasts ---
print("\nNew York Forecast Visualization")
ny_visualizer = GeohashForecastVisualizer(pipeline_ny.forecasts)
ny_visualizer.plot_all_geohash_forecasts(target_col="pm25")
ny_visualizer.plot_forecast_errors_distribution(target_col="pm25")

# Optionally choose a specific geohash
example_geohash_ny = pipeline_ny.forecasts['geohash'].unique()[0]
ny_visualizer.plot_forecast_for_geohash(example_geohash_ny, target_col="pm25")

print("\nBoston Forecast Visualization")
boston_visualizer = GeohashForecastVisualizer(pipeline_boston.forecasts)
boston_visualizer.plot_all_geohash_forecasts(target_col="pm25")
boston_visualizer.plot_forecast_errors_distribution(target_col="pm25")

example_geohash_boston = pipeline_boston.forecasts['geohash'].unique()[0]
boston_visualizer.plot_forecast_for_geohash(example_geohash_boston, target_col="pm25")
```
