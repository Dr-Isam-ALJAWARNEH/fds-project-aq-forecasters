```python
# 1. Imports
import sys, os
sys.path.append("/content/fds-project-aq-forecasters")
from scripts.preprocessing import AirQualityPreprocessor
import pandas as pd

# 2. Preprocess NEW YORK Data
# 2.1 Load multiple New York CSVs
data_path = "/content/fds-project-aq-forecasters/data/"
newyork_files = [
    data_path + "NYC_AQ/NYC_PM_Part1.csv",
    data_path + "NYC_AQ/NYC_PM_Part2.csv",
    data_path + "NYC_AQ/NYC_PM_Part3.csv"
]

# Load and concatenate
newyork_dfs = [pd.read_csv(file) for file in newyork_files]
newyork_combined = pd.concat(newyork_dfs, ignore_index=True)

# 2.2 Save temporarily (to use our Preprocessor easily)
temp_newyork_path = data_path + "NYC_AQ/newyork_combined_temp.csv"
newyork_combined.to_csv(temp_newyork_path, index=False)

# 2.3 Initialize preprocessor
preprocessor_ny = AirQualityPreprocessor(filepath=temp_newyork_path)

# 2.4 Run preprocessing steps
preprocessor_ny.load_data()
preprocessor_ny.basic_cleaning()
preprocessor_ny.add_geohash()
preprocessor_ny.handle_optional_columns()
preprocessor_ny.show_basic_info()

# 2.5 Save cleaned New York data
cleaned_dir = data_path + "cleaned"
os.makedirs(cleaned_dir, exist_ok=True)
cleaned_newyork_path = cleaned_dir + "/newyork_cleaned.csv"
preprocessor_ny.save_clean_data(cleaned_newyork_path)


# 3. Preprocess BOSTON Data
# 3.1 Boston file path
boston_file = data_path + "Boston_AQ/Boston_PM_NO2.csv"

# 3.2 Initialize preprocessor
preprocessor_boston = AirQualityPreprocessor(filepath=boston_file)

# 3.3 Run preprocessing steps
preprocessor_boston.load_data()
preprocessor_boston.basic_cleaning()
preprocessor_boston.add_geohash()
preprocessor_boston.handle_optional_columns()
preprocessor_boston.show_basic_info()

# 3.4 Save cleaned Boston data
cleaned_boston_path = cleaned_dir + "/boston_cleaned.csv"
preprocessor_boston.save_clean_data(cleaned_boston_path)

```
